## 💡 What is an AI Agent?

An **AI Agent** is a Large Language Model (LLM) that can:
- 🧠 **Think** — reason about what needs to be done
- 🛠️ **Use Tools** — call functions like search, calculator, APIs
- 👁️ **Observe** — look at tool results and decide next steps
- 🔄 **Repeat** — keep going until the task is done

```
User Question
     │
     ▼
  [THINK]     ← LLM decides what tool to call
     │
     ▼
  [CALL TOOL] ← Wikipedia / Tavily / add / multiply
     │
     ▼
  [OBSERVE]   ← Read tool result
     │
     ▼
 Done? ──No──► [THINK] again
     │Yes
     ▼
 Final Answer
```

## 🛠️ Step 3: Create Custom Tools with `@tool` Decorator

We create tools for:
1. **`add`**: Adds two numbers together.
2. **`multiply`**: Multiplies two numbers together.
3. **`wikipedia_search`**: Searches Wikipedia using the `wikipedia` library.

````python
from langchain_core.tools import tool
import wikipedia

# Set User-Agent for Wikipedia API compliance
wikipedia.set_user_agent("AIAgentTutorial/1.0 (contact@example.com)")

@tool
def add(a: float, b: float) -> float:
    """Add two numbers together.

    Args:
        a: The first number.
        b: The second number.

    Returns:
        The sum of a and b.
    """
    return a + b


@tool
def multiply(a: float, b: float) -> float:
    """Multiply two numbers together.

    Args:
        a: The first number.
        b: The second number.

    Returns:
        The product of a and b.
    """
    return a * b


@tool
def wikipedia_search(query: str) -> str:
    """Search Wikipedia for factual information, encyclopedic knowledge, biographies, and historical events.

    Args:
        query: The search term or topic to look up on Wikipedia.

    Returns:
        Summary text from Wikipedia.
    """
    try:
        results = wikipedia.search(query, results=3)
        if not results:
            return f"No Wikipedia pages found for '{query}'."
        page = wikipedia.page(results[0], auto_suggest=False)
        return f"Title: {page.title}\nSummary: {page.summary[:1500]}"
    except wikipedia.DisambiguationError as e:
        try:
            page = wikipedia.page(e.options[0], auto_suggest=False)
            return f"Title: {page.title}\nSummary: {page.summary[:1500]}"
        except Exception:
            return f"Multiple matches found for '{query}': {', '.join(e.options[:5])}"
    except Exception as e:
        return f"Error querying Wikipedia: {e}"

print("✅ Registered custom tools: add, multiply, wikipedia_search")
````

## 📚 Step 4: Configure Tavily Web Search Tool

- **`TavilySearchResults`**: Queries the live internet for recent news and real-time facts using your Tavily API key.

````python
from langchain_community.tools.tavily_search import TavilySearchResults

# Initialize Tavily Web Search Tool
tavily_tool = TavilySearchResults(
    max_results=3,
    description="Search the live web for recent events, breaking news, real-time data, and up-to-date information."
)

# Combine all tools into a single list
tools = [wikipedia_search, tavily_tool, add, multiply]

print(f"✅ Loaded {len(tools)} tools:")
for t in tools:
    print(f"  - {t.name}: {t.description[:70]}...")
````

## 🏗️ Step 5: Construct Agent with Plain Text Output & Tool Tracking

We configure the agent to output responses in clean, normal format and capture intermediate tool execution steps.

````python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
try:
    from langchain.agents import create_tool_calling_agent, AgentExecutor
except ImportError:
    from langchain_classic.agents import create_tool_calling_agent, AgentExecutor

# Define the System Prompt requesting normal plain text
prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        "You are a helpful, intelligent AI assistant equipped with specialized tools.\n"
        "You have access to:\n"
        "1. 'wikipedia_search' - for encyclopedic, historical, and conceptual knowledge from Wikipedia.\n"
        "2. 'tavily_search_results_json' - for live, real-time web searches and current news.\n"
        "3. 'add' - for adding two numbers precisely.\n"
        "4. 'multiply' - for multiplying two numbers precisely.\n\n"
        "Always choose the most appropriate tool for each sub-task. "
        "For arithmetic or calculations, always use the 'add' or 'multiply' tools rather than doing mental math. "
        "Output your final answers in clean, normal, easy-to-read plain text. "
        "Avoid markdown tables, markdown formatting symbols, and excessive hashes."
    ),
    MessagesPlaceholder(variable_name="chat_history", optional=True),
    ("human", "{input}"),
    MessagesPlaceholder(variable_name="agent_scratchpad"),
])

# Build the tool-calling agent
agent = create_tool_calling_agent(llm=llm, tools=tools, prompt=prompt)

# Create the Agent Executor with intermediate step tracking
agent_executor = AgentExecutor(
    agent=agent,
    tools=tools,
    verbose=False,
    return_intermediate_steps=True,
    handle_parsing_errors=True
)

# Runner function that displays tools used + normal plain text output
def ask_agent(user_input: str):
    result = agent_executor.invoke({"input": user_input})
    
    print("\n" + "=" * 60)
    print("USER QUERY:", user_input)
    print("-" * 60)
    
    # Check and print tools used
    intermediate_steps = result.get("intermediate_steps", [])
    if intermediate_steps:
        print("🛠️  TOOLS USED:")
        for action, _ in intermediate_steps:
            tool_name = action.tool
            tool_input = action.tool_input
            if tool_name == "wikipedia_search":
                print(f"  • Tool: [Wikipedia] -> Query: {tool_input}")
            elif tool_name == "tavily_search_results_json":
                print(f"  • Tool: [Tavily Web Search] -> Query: {tool_input}")
            elif tool_name == "add":
                print(f"  • Tool: [Add] -> Adding: {tool_input}")
            elif tool_name == "multiply":
                print(f"  • Tool: [Multiply] -> Multiplying: {tool_input}")
            else:
                print(f"  • Tool: [{tool_name}] -> Input: {tool_input}")
    else:
        print("🛠️  TOOLS USED: None (Direct Answer)")
        
    print("-" * 60)
    print("📄 OUTPUT:")
    print(result["output"])
    print("=" * 60 + "\n")

print("✅ AI Agent is ready!")
````

## 💬 Step 7: Interactive Chat Loop
Run the cell below to talk directly with your AI Agent.

````python
print("🤖 AI Agent interactive chat session started!")
print("Type 'exit' or 'quit' to end.\n")

while True:
    user_input = input("\n👤 You: ").strip()
    if not user_input:
        continue
    if user_input.lower() in ["exit", "quit", "q"]:
        print("\n👋 Goodbye!")
        break
    
    try:
        ask_agent(user_input)
    except Exception as err:
        print(f"\n❌ Error: {err}")
````
